Q1. Who are the most listened‑to artists vs less listened‑to artists?

```python
artist_streams = (
    df.groupby("Artist", observed=True)["Stream"]
      .sum()
      .sort_values(ascending=False)
)

artist_streams.head(10)   # top 10 most streamed artists
artist_streams.tail(10)   # 10 least streamed artists (among those with >=1 track)
```

 top 10 artists by total streams

```python
import matplotlib.pyplot as plt
import seaborn as sns

top_10_artists = artist_streams.head(10).index

top10_df = df[df["Artist"].isin(top_10_artists)]

artist_total = (
    top10_df.groupby("Artist", observed=True)["Stream"]
            .sum()
            .reindex(top_10_artists)
)

artist_total.plot(kind="barh", figsize=(8, 6))
plt.xlabel("Total streams")
plt.ylabel("Artist")
plt.title("Top 10 most streamed artists")
plt.tight_layout()
plt.show()
```

 Distribution of streams per artist 
 

```python
artist_track_counts = df.groupby("Artist", observed=True).size()
artist_track_counts.describe()
```

```python
# How many tracks per artist?
artist_track_counts.hist(bins=50)
plt.xlabel("Number of tracks per artist")
plt.ylabel("Number of artists")
plt.title("Distribution of tracks per artist")
plt.tight_layout()
plt.show()
```

```python
artist_avg = (
    df.groupby("Artist", observed=True)["Stream"]
      .mean()
      .sort_values(ascending=False)
)

artist_avg.head(10)  # highest average streams per track
artist_avg.tail(10)  # lowest average streams per track
```

Q2. What makes less‑listened‑to artists different from the biggest ones?

```python
# Total streams per artist
artist_total_all = (
    df.groupby("Artist", observed=True)["Stream"]
      .sum()
)

median_streams = artist_total_all.median()

# Map each track to its artist's total streams
df["artist_total_streams"] = df["Artist"].map(artist_total_all)

# Label tracks as from "big" or "small" artists
df["artist_size"] = (
    df["artist_total_streams"]
      .apply(lambda x: "big" if x >= median_streams else "small")
)

df["artist_size"].value_counts()
```

Compare audio features: big vs small artists

```python
features = ["Danceability", "Energy", "Loudness", "Speechiness",
            "Acousticness", "Valence", "Tempo", "Duration_min"]
```

```python
df.groupby("artist_size")[features].mean().T
```

```python
plt.figure(figsize=(10, 6))
sns.boxplot(data=df, x="artist_size", y="Energy", order=["small", "big"])
plt.title("Energy: small vs big artists")
plt.xlabel("Artist size")
plt.ylabel("Energy")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.boxplot(data=df, x="artist_size", y="Danceability", order=["small", "big"])
plt.title("Danceability: small vs big artists")
plt.xlabel("Artist size")
plt.ylabel("Danceability")
plt.tight_layout()
plt.show()
```

Q3. Between licensed and unlicensed artists, which ones are getting more streams?


```python
df["Licensed"].value_counts()
df["Licensed"].dtype
```

```python
df.groupby("Licensed", observed=True)["log_Stream"].describe()
```

```python
plt.figure(figsize=(8, 6))
sns.boxplot(data=df, x="Licensed", y="log_Stream", order=[False, True])
plt.xlabel("Licensed")
plt.ylabel("Log(1 + Stream)")
plt.title("Streams: licensed vs unlicensed")
plt.tight_layout()
plt.show()
```

```python
avg_by_licensed = (
    df.groupby("Licensed", observed=True)["log_Stream"]
      .mean()
)

avg_by_licensed.plot(kind="bar")
plt.xlabel("Licensed")
plt.ylabel("Average log(1 + Stream)")
plt.title("Average streams by licensing status")
plt.tight_layout()
plt.show()
```
